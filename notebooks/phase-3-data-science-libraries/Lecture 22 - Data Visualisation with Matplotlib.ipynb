### The Pyplot Interface: Basic Plot Types

Matplotlib's `pyplot` module provides a MATLAB-like interface for creating plots. `plt.plot()` draws line plots, `plt.bar()` creates bar charts, `plt.hist()` generates histograms, and `plt.scatter()` produces scatter plots. Each function accepts data as the primary arguments and optional keyword arguments for customisation.

Starting with `plt.figure()` creates a new figure, and subsequent plot commands draw on it. `plt.show()` renders the figure. This procedural interface is great for quick exploratory plots, though the object-oriented interface (using Figure and Axes directly) gives more control for complex layouts.

```python
import matplotlib.pyplot as plt
import numpy as np

# Line plot
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, label='sin(x)', color='blue', linewidth=2)
plt.plot(x, np.cos(x), 'r--', label='cos(x)', linewidth=2)
plt.title('Sine and Cosine Waves')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Bar chart, histogram, and scatter
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 33]

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.bar(categories, values, color='skyblue', edgecolor='navy')
plt.title('Bar Chart')

plt.subplot(1, 3, 2)
data = np.random.randn(1000)
plt.hist(data, bins=30, color='green', alpha=0.7, edgecolor='black')
plt.title('Histogram')

plt.subplot(1, 3, 3)
x = np.random.randn(200)
y = x + np.random.randn(200) * 0.5
plt.scatter(x, y, alpha=0.5, c='purple', s=20)
plt.title('Scatter Plot')

plt.tight_layout()
plt.show()
```

### Figure and Axes Objects: The Object-Oriented Interface

The object-oriented interface gives you explicit control over the figure and its axes. `plt.subplots()` creates a Figure and one or more Axes objects at once. You then call methods on the Axes objects (like `ax.plot()`, `ax.set_title()`) rather than using `plt.*` functions.

This approach is essential for complex layouts. You can create a grid of subplots with a single call and access each Axes individually to customise titles, labels, and data. It also makes it easier to reuse plotting code inside functions.

### Customisation and Styling

Matplotlib offers extensive customisation options. You can control colours (named, hex, RGB), markers (`.`, `o`, `s`, `^`), line styles (`-`, `--`, `-.`, `:`), and transparency with `alpha`. The `plt.style.use()` function lets you apply predefined style sheets like `'seaborn-v0_8'`, `'ggplot'`, or `'fivethirtyeight'` for instantly better-looking plots.

Axis limits can be set with `xlim()` / `ylim()` or `ax.set_xlim()`. Legends are positioned with `loc` and can be placed outside the plot area. `plt.tight_layout()` automatically adjusts spacing to prevent overlapping elements. These customisations turn basic plots into publication-quality figures.

```python
# Styling with style sheets
plt.style.use('seaborn-v0_8-darkgrid')

# Custom styling: colours, markers, line styles
x = np.linspace(0, 4 * np.pi, 100)

plt.figure(figsize=(10, 6))
plt.plot(x, np.sin(x), color='#E74C3C', marker='o', markersize=4,
         linestyle='-', linewidth=2, label='sin(x)')
plt.plot(x, np.sin(x + 1), color='#2ECC71', marker='s', markersize=4,
         linestyle='--', linewidth=2, label='sin(x+1)')
plt.plot(x, np.sin(x + 2), color='#3498DB', marker='^', markersize=4,
         linestyle=':', linewidth=2, label='sin(x+2)')

plt.title('Custom Styling Demo', fontsize=14, fontweight='bold')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.legend(loc='upper right', frameon=True, shadow=True)
plt.xlim(0, 4 * np.pi)
plt.tight_layout()
plt.show()
```

```python
# Saving figures
fig, ax = plt.subplots(figsize=(8, 4))
x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x) * np.exp(-x/5), 'b-', linewidth=2)
ax.set_title('Damped Oscillation')
ax.set_xlabel('Time')
ax.set_ylabel('Amplitude')
ax.grid(True, alpha=0.3)
plt.tight_layout()

# Save to file
plt.savefig('damped_oscillation.png', dpi=150, bbox_inches='tight')
print('Figure saved as damped_oscillation.png')
```
