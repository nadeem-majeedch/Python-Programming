### Reshaping, Transposing, Stacking, and Splitting

Manipulating array shapes is a common task in data preparation. `reshape()` changes an array's dimensions without altering its data (the total number of elements must stay the same). `T` or `transpose()` flips axes, turning rows into columns. `stack()` joins arrays along a new axis, while `concatenate()` joins along an existing axis. `split()` does the reverse, dividing an array into multiple sub-arrays.

These operations are essential when preparing data for machine learning models, which often expect specific input shapes. For example, you might stack feature arrays column-wise or split a dataset into training and validation folds.

```python
# Reshaping and transposing
arr = np.arange(12)
print('Original:', arr)

reshaped = arr.reshape(3, 4)
print('Reshaped (3x4):\n', reshaped)

transposed = reshaped.T
print('Transposed:\n', transposed)
```

```python
# Stacking and splitting
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print('Vertical stack:\n', np.vstack([a, b]))
print('Horizontal stack:\n', np.hstack([a, b]))
print('Column stack:\n', np.column_stack([a, b]))

# Splitting
arr = np.arange(10)
first, second, third = np.split(arr, [3, 7])
print('Split result:', first, second, third)
```

### Matrix Multiplication, Inverses, and Eigenvalues

Linear algebra is at the heart of many data science algorithms, from linear regression to PCA and neural networks. NumPy's `linalg` module provides all the operations you need. `np.dot(a, b)` and the `@` operator both compute matrix products. `np.linalg.inv()` computes the inverse of a square matrix, and `np.linalg.eig()` returns eigenvalues and eigenvectors.

The `@` operator (Python 3.5+) is the most readable way to express matrix multiplication. For a linear regression solution, the normal equation `w = (X^T X)^{-1} X^T y` can be written in a single line of readable code.

```python
# Matrix multiplication with @ and np.dot
X = np.array([[1, 2], [3, 4]])
y = np.array([5, 6])

print('X @ y:', X @ y)
print('np.dot(X, y):', np.dot(X, y))
```

```python
# Linear algebra: matrix inverse and linear regression
np.random.seed(42)
X = np.random.randn(100, 3)
true_w = np.array([2.0, -1.5, 0.5])
y = X @ true_w + np.random.randn(100) * 0.2

# Normal equation: w = (X^T X)^{-1} X^T y
XTX_inv = np.linalg.inv(X.T @ X)
w_hat = XTX_inv @ X.T @ y

print('True weights:', true_w)
print('Estimated w:', np.round(w_hat, 3))

# Eigenvalues
cov_matrix = np.cov(X.T)
eigvals, eigvecs = np.linalg.eig(cov_matrix)
print('Eigenvalues:', np.round(eigvals, 3))
```

### The Random Module: Generating Random Numbers

NumPy's `np.random` module provides functions to generate random numbers from various distributions. `np.random.randn()` draws from the standard normal distribution, `.uniform()` from a uniform distribution, and `.choice()` randomly samples from an array. You can also generate random integers with `.randint()`.

Reproducibility is critical in data science. Setting `np.random.seed(seed_value)` initialises the random number generator to a deterministic state. Any code run after setting the seed will produce the same random numbers, making experiments reproducible across runs and across different machines.

```python
# Random number generation
np.random.seed(42)

# Normal distribution
normal_sample = np.random.randn(1000)
print('Normal sample mean:', np.mean(normal_sample).round(3))
print('Normal sample std:', np.std(normal_sample).round(3))

# Uniform distribution
uniform_sample = np.random.uniform(0, 10, 1000)
print('Uniform sample mean:', np.mean(uniform_sample).round(3))

# Random choice
choices = np.random.choice(['red', 'blue', 'green'], size=10, p=[0.5, 0.3, 0.2])
print('Weighted choices:', choices)
```

```python
# Monte Carlo: estimating pi
np.random.seed(123)
n_points = 1_000_000

# Random points in unit square [-1, 1] x [-1, 1]
x = np.random.uniform(-1, 1, n_points)
y = np.random.uniform(-1, 1, n_points)

# Count points inside unit circle
inside = (x**2 + y**2) <= 1
pi_estimate = 4 * inside.sum() / n_points

print(f'Estimated pi: {pi_estimate:.5f}')
print(f'Actual pi:    {np.pi:.5f}')
```

### Performance Comparison: Loops vs Vectorised NumPy

Python loops are interpreted one iteration at a time, which incurs significant overhead for each operation. NumPy's vectorised operations delegate the iteration to highly optimised C and Fortran libraries (BLAS/LAPACK). The speed difference grows with data size; for large arrays, vectorised code can be 50-100x faster than equivalent loops.

Beyond speed, vectorised code is shorter, more readable, and less error-prone. A rule of thumb: if you find yourself writing a `for` loop over array elements, there is probably a faster NumPy alternative.

```python
# Performance comparison: Python loops vs vectorised NumPy
import time

n = 5_000_000
a = np.random.randn(n)
b = np.random.randn(n)

# Python loop
start = time.time()
result_loop = [a[i] * b[i] + a[i] for i in range(n)]
loop_time = time.time() - start
print(f'Python loop: {loop_time:.3f}s')

# Vectorised
start = time.time()
result_np = a * b + a
np_time = time.time() - start
print(f'NumPy:       {np_time:.3f}s')
print(f'Speedup:     {loop_time / np_time:.1f}x')
```
