### Setting the Theme and Distribution Plots

Seaborn extends Matplotlib with statistically meaningful visualisations and attractive default styles. `sns.set_theme()` applies a unified style that instantly improves the look of all plots. `sns.set_palette()` controls the colour palette, with options like `'husl'`, `'Set2'`, `'viridis'`, and `'coolwarm'`.

Distribution plots reveal the shape, spread, and outliers of your data. `histplot()` creates histograms with optional KDE overlay. `kdeplot()` shows the estimated probability density. `boxplot()` displays quartiles and outliers, while `violinplot()` combines box plot and KDE for a richer picture.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Set theme and palette
sns.set_theme(style='darkgrid')
sns.set_palette('husl')

# Generate data
np.random.seed(42)
data = pd.DataFrame({
    'value': np.random.randn(500),
    'group': np.random.choice(['A', 'B', 'C'], 500)
})

# Distribution plots
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

sns.histplot(data['value'], bins=30, kde=True, ax=axes[0, 0])
axes[0, 0].set_title('histplot with KDE')

sns.kdeplot(data['value'], fill=True, ax=axes[0, 1])
axes[0, 1].set_title('kdeplot')

sns.boxplot(x='group', y='value', data=data, ax=axes[1, 0])
axes[1, 0].set_title('boxplot')

sns.violinplot(x='group', y='value', data=data, ax=axes[1, 1])
axes[1, 1].set_title('violinplot')

plt.tight_layout()
plt.show()
```

### Relationship and Categorical Plots

Seaborn provides specialised functions for exploring relationships between variables. `scatterplot()` is the basic scatter plot. `relplot()` is a figure-level function that creates scatter or line plots with optional faceting by additional variables. `lmplot()` adds a regression line with confidence bands, making it easy to visualise linear trends.

For categorical data, `countplot()` shows the count of observations in each category, similar to a bar chart of value counts. `barplot()` shows the mean (or other estimator) of a numeric variable for each category, with error bars. `catplot()` is the figure-level version that supports multiple categorical plot types and faceting.

```python
# Relationship plots
df = pd.DataFrame({
    'hours_studied': np.random.uniform(1, 20, 200),
    'exam_score': np.random.uniform(1, 20, 200) * 4 + np.random.randn(200) * 10,
    'passed': np.random.choice(['Yes', 'No'], 200, p=[0.7, 0.3])
})

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.scatterplot(x='hours_studied', y='exam_score', data=df, ax=axes[0])
axes[0].set_title('scatterplot')

sns.scatterplot(x='hours_studied', y='exam_score', hue='passed',
                data=df, alpha=0.6, ax=axes[1])
axes[1].set_title('scatterplot with hue')

sns.lmplot(x='hours_studied', y='exam_score', data=df, ax=axes[2])
axes[2].set_title('lmplot with regression line')

plt.tight_layout()
plt.show()
```

```python
# Categorical plots
tips = sns.load_dataset('tips')
print('Tips dataset loaded:', tips.shape)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.countplot(x='day', data=tips, ax=axes[0])
axes[0].set_title('countplot - orders by day')

sns.barplot(x='day', y='total_bill', data=tips, ax=axes[1])
axes[1].set_title('barplot - avg bill by day')

sns.catplot(x='day', y='total_bill', hue='sex', kind='box',
            data=tips, ax=axes[2])
axes[2].set_title('catplot - box by day/sex')

plt.tight_layout()
plt.show()
```

### Pair Plots, Heatmaps, and FacetGrids

`sns.pairplot()` creates a grid of scatter plots and histograms for every pair of columns in a DataFrame. It is the single most useful function for initial multivariate exploration. The diagonal shows distributions, and off-diagonal cells show pairwise relationships. Adding `hue` colour-codes the scatter plots by a categorical variable.

`sns.heatmap()` visualises a matrix (typically a correlation matrix) with colour coding, making it easy to spot strong positive or negative relationships. `FacetGrid` is a powerful tool for creating multi-panel plots conditioned on one or more categorical variables, allowing you to compare distributions across subgroups.

```python
# Pair plot
iris = sns.load_dataset('iris')
print('Iris dataset loaded:', iris.shape)

sns.pairplot(iris, hue='species', diag_kind='kde',
             palette='Set2', height=2.5)
plt.suptitle('Pair Plot of Iris Dataset', y=1.02)
plt.show()
```

```python
# Heatmap for correlation
corr = iris.select_dtypes(include='number').corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=1, fmt='.2f')
plt.title('Correlation Heatmap - Iris Dataset')
plt.show()
```

```python
# FacetGrid for multi-panel plots
g = sns.FacetGrid(tips, col='time', row='sex', margin_titles=True)
g.map(sns.histplot, 'total_bill', bins=20)
g.fig.suptitle('Total Bill Distribution by Time and Gender', y=1.02)
plt.show()
```

## Data Science Connection

Statistical visualisation is the bridge between raw data and actionable insights. Distribution plots reveal data quality issues (skewness, outliers, multimodality). Relationship plots suggest hypotheses about cause and effect. Pair plots and heatmaps provide a bird's-eye view of multivariate structure. Seaborn makes these analyses accessible with concise, high-level functions that produce publication-quality figures.
