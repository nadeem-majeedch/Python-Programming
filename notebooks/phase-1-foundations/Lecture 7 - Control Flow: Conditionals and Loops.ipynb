## Boolean Expressions and Truthiness

Every value in Python has an inherent truth value, known as **truthiness**. The following are considered `False`: `None`, `False`, zero numeric values (`0`, `0.0`, `0j`), empty sequences (`''`, `[]`, `()`), and empty mappings (`{}`). Everything else is `True`. This means you can use any value directly in a conditional: `if my_list:` is equivalent to `if len(my_list) > 0:`.

Boolean expressions can be combined with `and`, `or`, and `not`. Python uses **short-circuit evaluation**: in `a and b`, if `a` is `False`, `b` is never evaluated. In `a or b`, if `a` is `True`, `b` is never evaluated. This is both an optimization and a useful programming pattern.

```python
# Truthiness of different values
values = [None, False, 0, 0.0, '', [], {}, 'hello', [1, 2], {'a': 1}, 42]
for val in values:
    print(f"bool({val!r:8}) = {bool(val)}")

# Using truthiness in conditionals
name = ""
if name:
    print(f"Hello, {name}")
else:
    print("Name is empty!")
```

```python
# Short-circuit evaluation
# In 'and', if first is False, second is not evaluated
def risky_operation():
    print("Risky operation executed!")
    return True

result = False and risky_operation()
print("Result of False and risky():", result)

# Practical: safe attribute access
data = {"name": "Alice"}
if 'age' in data and data['age'] > 18:
    print("Adult")
else:
    print("Age unknown or under 18")
```

## if/elif/else Chains

Conditional execution is the backbone of decision-making in code. The `if` statement executes a block if its condition is `True`. You can chain additional conditions with `elif` (short for 'else if'), and provide a fallback with `else`. Only one branch in the chain executes — the first one whose condition is `True`.

In data science, conditionals are used everywhere: filtering data, assigning categories, handling missing values, and implementing business logic. For example, you might classify a customer as 'High', 'Medium', or 'Low' value based on their spending, or flag outliers based on z-score thresholds.

```python
# Basic if/elif/else
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'
print(f"Score: {score}, Grade: {grade}")
```

```python
# Conditional logic for data filtering
transactions = [
    {"amount": 250, "type": "purchase"},
    {"amount": 5000, "type": "purchase"},
    {"amount": 150, "type": "refund"},
    {"amount": 12000, "type": "purchase"},
    {"amount": 75, "type": "purchase"}
]

for txn in transactions:
    if txn['amount'] > 10000:
        txn['flag'] = 'large_txn'
    elif txn['type'] == 'refund':
        txn['flag'] = 'refund'
    elif txn['amount'] < 100:
        txn['flag'] = 'small_txn'
    else:
        txn['flag'] = 'normal'

for txn in transactions:
    print(txn)
```

## for Loops Over Lists, Strings, Dicts, and Ranges

The `for` loop is the primary tool for iteration in Python. It works with any **iterable** — lists, strings, dictionaries, sets, tuples, files, and more. The loop variable takes each element in turn, and the loop body executes once per element.

Iterating over different data structures follows slightly different patterns: over a list, you get each element; over a string, you get each character; over a dictionary, by default you get each key (use `.items()` to get key-value pairs). The `range()` function generates sequences of numbers and is commonly used when you need to iterate a specific number of times or when you need index-based access.

```python
# for loops over different data structures
# Over a list
fruits = ["apple", "banana", "cherry"]
print("Fruits:")
for fruit in fruits:
    print(f"  - {fruit}")

# Over a string (character by character)
print("\nCharacters:")
for char in "Python":
    print(f"  '{char}'")

# Over a dictionary
ages = {"Alice": 28, "Bob": 35, "Charlie": 22}
print("\nAges:")
for name, age in ages.items():
    print(f"  {name} is {age}")
```

```python
# for loop with range()
print("Range 0 to 4:")
for i in range(5):
    print(f"  {i}")

print("\nRange 2 to 7:")
for i in range(2, 8):
    print(f"  {i}")

print("\nRange 10 to 1 step -2:")
for i in range(10, 0, -2):
    print(f"  {i}")

# Practical: enumerate with index
colors = ["red", "green", "blue"]
print("\nEnumerated colors:")
for idx, color in enumerate(colors):
    print(f"  {idx}: {color}")
```

## The range() Function

`range(start, stop, step)` generates an immutable sequence of numbers. It is lazy — it does not create the entire list in memory, making it memory-efficient even for very large ranges. `range(stop)` starts at 0 and goes up to (but not including) `stop`. `range(start, stop)` starts at `start`. `range(start, stop, step)` increments by `step` (which can be negative).

In data science, `range()` is commonly used for: creating index sequences for iteration, generating bins for histograms, creating evenly spaced values for plotting, and batch processing large datasets in chunks.

```python
# range() variations
# Basic range from 0 to stop-1
print("list(range(5)):", list(range(5)))

# Range with start and stop
print("list(range(2, 8)):", list(range(2, 8)))

# Range with step
print("list(range(0, 20, 3)):", list(range(0, 20, 3)))

# Negative step (reverse)
print("list(range(10, 0, -2)):", list(range(10, 0, -2)))
```

```python
# Practical use of range() in data science
# Simulating batch processing
total_rows = 100
batch_size = 30

for batch_start in range(0, total_rows, batch_size):
    batch_end = min(batch_start + batch_size, total_rows)
    print(f"Processing rows {batch_start} to {batch_end}")

# Creating evenly spaced points for plotting
num_points = 10
spacing = 0.5
points = [i * spacing for i in range(num_points)]
print("Data points:", points)
```

## while Loops and Infinite Loop Prevention

A `while` loop repeats as long as its condition is `True`. It is used when you do not know in advance how many iterations you need — for example, reading from a file until EOF, or waiting for a condition to change. Because the loop checks the condition only at the start of each iteration, it is possible to write an **infinite loop** if the condition never becomes `False`.

To prevent infinite loops, always ensure that something changes inside the loop body that will eventually make the condition `False`. Common strategies include: incrementing a counter, removing elements from a collection, or reading input until a sentinel value is received. A safety break after a maximum number of iterations is also a good practice.

```python
# Basic while loop with counter
count = 0
while count < 5:
    print(f"Iteration {count}")
    count += 1
print("Loop finished!")
```

```python
# While loop with sentinel value
# Simulating reading data until a termination signal
data_stream = [10, 20, 30, -1, 40, 50]
idx = 0
results = []

while idx < len(data_stream) and data_stream[idx] != -1:
    results.append(data_stream[idx] * 2)
    idx += 1
print("Processed values:", results)
print(f"Stopped at index {idx} (value was {data_stream[idx]})")

# Safety max-iterations guard
max_iter = 1000
iteration = 0
while iteration < max_iter:
    # Simulated processing
    if iteration >= 3:
        break  # normal exit
    iteration += 1
```

## break, continue, and the Loop else Clause

`break` immediately exits the loop, skipping any remaining iterations. It is commonly used to stop searching once a target is found. `continue` skips the rest of the current iteration and moves to the next one. It is useful for filtering — you can skip invalid items and process only valid ones.

The `else` clause on a loop is a Python feature that is often misunderstood. The `else` block executes **only if the loop completed normally** (i.e., without hitting a `break`). It does **not** execute if the loop was exited via `break`. This is useful for search scenarios: if you loop through items looking for something and `break` when found, the `else` block runs if the item was never found.

```python
# break: exit loop early
numbers = [3, 7, 2, 9, 4, 6]
target = 9
for num in numbers:
    if num == target:
        print(f"Found {target} at index {numbers.index(target)}")
        break
else:
    print(f"{target} not found")

# Without break, else would execute
for num in [1, 2, 3]:
    print(f"Checking {num}")
else:
    print("Loop completed without break")
```

```python
# continue: skip invalid data
raw_data = [10, -5, 20, None, 30, -1, 40, 0]
valid = []
for value in raw_data:
    if value is None or value < 0:
        continue  # skip invalid
    valid.append(value * 2)
print("Original:", raw_data)
print("Processed valid:", valid)

# break + else for searching
def find_value(data, target):
    for item in data:
        if item == target:
            print(f"Found {target}")
            break
    else:
        print(f"{target} not found")

find_value([1, 2, 3, 4, 5], 3)
find_value([1, 2, 3, 4, 5], 99)
```

The `break`/`continue`/`else` patterns you learned here appear in real-world code for tasks like early stopping during model training (break when validation loss stops improving), skipping corrupted files during data ingestion (continue), and validating that all required columns exist before processing (loop-else).
