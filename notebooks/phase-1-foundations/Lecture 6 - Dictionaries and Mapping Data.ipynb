## Key-Value Pairs and Hashability Constraints

A dictionary (or `dict`) is an unordered collection of key-value pairs, where each key maps to a value. Dictionaries are created with curly braces `{}` using the syntax `{key: value, ...}`. Keys must be **hashable** (immutable types like strings, numbers, and tuples), while values can be any Python object — including lists, other dictionaries, or even functions.

Dictionaries are the closest native Python data structure to a database record or a JSON object. In data science, you will use dictionaries constantly to represent individual data records, store configuration parameters, build lookup tables, and accumulate group-wise statistics. Their O(1) average lookup time makes them extremely efficient for mapping operations.

```python
# Creating dictionaries
empty = {}
person = {
    "name": "Alice",
    'age': 28,
    'city': 'New York',
    'skills': ['Python', 'SQL', 'Statistics']
}
print("Person:", person)
print("Name:", person["name"])

# Keys must be hashable — tuples work, lists do not
locations = {
    (40.71, -74.01): "NYC",
    (34.05, -118.24): "LA"
}
print("Location:", locations[(40.71, -74.01)])
```

```python
# Accessing and modifying dictionaries
student = {"id": 101, "name": "Bob", "grade": 85}
print("Original:", student)

# Access with square brackets
print("Grade:", student["grade"])

# Modify existing key
student["grade"] = 90
print("After grade update:", student)

# Add new key-value pair
student["passed"] = True
print("After adding passed:", student)

# KeyError if key does not exist
# print(student["gpa"])  # KeyError
```

## Dictionary Methods: .keys(), .values(), .items(), .get(), .setdefault()

Python dictionaries provide methods that make common operations convenient: `.keys()` returns a view of all keys, `.values()` returns a view of all values, and `.items()` returns a view of all key-value pairs as tuples. These views are dynamic — they reflect changes to the dictionary in real time.

The `.get()` method is a safer alternative to bracket access: it returns `None` (or a default you provide) if the key is missing, instead of raising a `KeyError`. The `.setdefault()` method goes a step further: it returns the value if the key exists, and if not, inserts the key with a default value and returns that. Both are invaluable for building counting and aggregation logic.

```python
# .keys(), .values(), .items()
inventory = {"apples": 10, "bananas": 5, "cherries": 20}
print("Keys:", list(inventory.keys()))
print("Values:", list(inventory.values()))
print("Items:", list(inventory.items()))

# Iterating over items
for item, quantity in inventory.items():
    print(f"{item}: {quantity}")
```

```python
# .get() for safe access
scores = {"Alice": 95, "Bob": 87}
print("Alice:", scores.get("Alice"))
print("Charlie:", scores.get("Charlie"))
print("Charlie with default:", scores.get("Charlie", 0))

# .setdefault() — get or insert
visits = {}
users = ["Alice", "Bob", "Alice", "Charlie", "Bob", "Alice"]
for user in users:
    visits.setdefault(user, 0)
    visits[user] += 1
print("Visit counts:", visits)
```

## Iterating Over Dictionaries

You can iterate over a dictionary directly in a `for` loop, which yields keys by default. For most cases, you will want to iterate over `.items()` to get both the key and value in each iteration. You can also iterate over `.keys()` or `.values()` when you only need one side.

Dictionary iteration is fundamental to data aggregation tasks. For example, you might iterate over a list of raw records and build a dictionary that groups values by a category key. This pattern — "iterate and accumulate into a dict" — appears in virtually every data science workflow.

```python
# Different ways to iterate
grades = {"Alice": [85, 90, 92], "Bob": [78, 81, 85], "Charlie": [92, 95, 98]}

# Iterate over keys (default)
print("Students:")
for student in grades:
    print(f"  - {student}")

# Iterate over items
print("\nAverages:")
for student, score_list in grades.items():
    avg = sum(score_list) / len(score_list)
    print(f"  {student}: {avg:.1f}")
```

```python
# Building an aggregation dictionary
sales_data = [
    ("NY", 100), ("CA", 200), ("NY", 150),
    ("TX", 300), ("CA", 250), ("NY", 50)
]

# Aggregate sales by state
state_sales = {}
for state, amount in sales_data:
    if state in state_sales:
        state_sales[state] += amount
    else:
        state_sales[state] = amount
print("Sales by state:", state_sales)

# Same thing using .get()
state_sales2 = {}
for state, amount in sales_data:
    state_sales2[state] = state_sales2.get(state, 0) + amount
print("Sales by state (v2):", state_sales2)
```

## Dictionary Comprehensions

Just like list comprehensions, Python supports dictionary comprehensions with the syntax `{key_expr: value_expr for item in iterable}`. They provide a concise way to build dictionaries from existing iterables. You can also add filtering conditions with `if` clauses.

Dictionary comprehensions are widely used in data science for: transforming one mapping into another (e.g., converting column names to lowercase), creating lookup tables from lists, and filtering dictionaries by key or value conditions.

```python
# Basic dictionary comprehension
squares = {x: x**2 for x in range(1, 6)}
print("Squares:", squares)

# From two lists (zipping)
cities = ["NYC", "LA", "Chicago"]
populations = [8336817, 3898747, 2746388]
city_pop = {city: pop for city, pop in zip(cities, populations)}
print("City populations:", city_pop)

# Filtering with a condition
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print("Even squares:", even_squares)
```

```python
# Dictionary comprehension in data science context
# Transform column names to lowercase
raw_columns = ["Name", "Age ", " Salary", "City"]
clean_columns = {col: col.strip().lower() for col in raw_columns}
print("Cleaned mapping:", clean_columns)

# Invert a dictionary (swap keys and values)
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print("Inverted:", inverted)

# Filter by value
scores = {"Alice": 95, "Bob": 72, "Charlie": 88, "Diana": 60}
passing = {name: score for name, score in scores.items() if score >= 70}
print("Passing students:", passing)
```

## defaultdict and Counter from collections

The `collections` module provides specialized dictionary classes. `defaultdict` is a dictionary that supplies a default value for missing keys automatically, eliminating the need for explicit checks. You provide a factory function (like `int`, `list`, `float`) when creating it.

`Counter` is a dictionary subclass designed for counting hashable objects. It takes an iterable and automatically counts occurrences. It also provides helpful methods like `.most_common()` to get the most frequent items. Both `defaultdict` and `Counter` are indispensable tools in the data scientist's toolkit.

```python
# defaultdict for clean aggregation
from collections import defaultdict, Counter

# Without defaultdict (verbose)
word_counts = {}
text = "the cat in the hat the cat sat".split()
for word in text:
    if word not in word_counts:
        word_counts[word] = 0
    word_counts[word] += 1
print("Manual count:", word_counts)

# With defaultdict (clean)
word_counts2 = defaultdict(int)
for word in text:
    word_counts2[word] += 1
print("DefaultDict count:", dict(word_counts2))
```

```python
# defaultdict with list factory
from collections import defaultdict

# Group items by category
products = [
    ("fruit", "apple"), ("fruit", "banana"),
    ("veg", "carrot"), ("fruit", "cherry"),
    ("veg", "broccoli"), ("dairy", "milk")
]

by_category = defaultdict(list)
for category, product in products:
    by_category[category].append(product)
print("Grouped products:", dict(by_category))

# Access a missing key returns empty list
print("Meat category:", by_category["meat"])
```

```python
# Counter for frequency analysis
from collections import Counter

# Survey responses
responses = [
    "Python", "R", "Python", "Python", "Julia",
    "R", "Python", "JavaScript", "Julia", "Python", "R"
]

lang_counts = Counter(responses)
print("Language counts:", lang_counts)

# Most common
print("Most common:", lang_counts.most_common(2))

# Counter operations
more_responses = ["Python", "Rust", "Rust", "Python"]
lang_counts.update(more_responses)
print("After update:", lang_counts)

# Count characters in a string
text = "hello world"
char_count = Counter(text)
print("Character frequencies:", char_count)
```
