## Tuple Creation, Immutability, and Unpacking

A tuple is an ordered, immutable collection of elements, written with parentheses `()`. Once created, you cannot add, remove, or change elements. This immutability makes tuples **hashable** (if all their elements are also hashable), which means they can be used as dictionary keys and members of sets — something lists cannot do.

**Tuple unpacking** is a powerful feature: you can assign each element of a tuple to a separate variable in one line: `name, age, city = person`. This is extensively used in data science for iterating over collections of structured data, returning multiple values from functions, and swapping variables elegantly.

```python
# Creating tuples
empty = ()
single = (42,)  # trailing comma is REQUIRED for single-element tuple
coordinates = (40.7128, -74.0060)
person = ("Alice", 28, "Engineer")
print("Coordinates:", coordinates)
print("Person:", person)

# Tuples are immutable
# coordinates[0] = 0  # TypeError

# Tuples can be used as dictionary keys
location_map = {(40.71, -74.01): "NYC", (34.05, -118.24): "LA"}
print("Location map:", location_map)
```

```python
# Tuple unpacking
record = ("Bob", 35, 95000)
name, age, salary = record
print(f"Name: {name}, Age: {age}, Salary: ${salary}")

# Swapping variables with tuples
a, b = 10, 20
print(f"Before swap: a={a}, b={b}")
a, b = b, a
print(f"After swap: a={a}, b={b}")

# Unpacking in a for loop (common in data science)
employees = [
    ("Alice", 28, 75000),
    ("Bob", 35, 95000),
    ("Charlie", 22, 45000)
]
for name, age, salary in employees:
    print(f"{name} is {age} years old, earns ${salary}")
```

## When to Prefer a Tuple Over a List

Use a tuple when: (1) the data should not change (immutability provides safety), (2) you need to use the collection as a dictionary key, (3) you want to return multiple values from a function, or (4) you want to communicate to other developers that the sequence is fixed. Tuples are also slightly more memory-efficient than lists.

In data science, tuples are commonly used for: coordinates (lat, lon), RGB color values, function return values, dictionary keys for composite identifiers (e.g., `(user_id, date)`), and fixed configuration parameters that should not be changed accidentally.

```python
# Tuple as a dictionary key (composite key)
sales = {}
sales[("NY", "2024-01")] = 15000
sales[("CA", "2024-01")] = 22000
sales[("NY", "2024-02")] = 18000
print("NY Jan sales:", sales[("NY", "2024-01")])

# Tuple for fixed configuration
COLORS = [
    ("red", "#FF0000"),
    ("green", "#00FF00"),
    ("blue", "#0000FF")
]
for name, hex_code in COLORS:
    print(f"{name}: {hex_code}")
```

```python
# Tuple vs list: immutability as a safety feature
# If you pass a list to a function, it can be modified
def add_item_buggy(items):
    items.append('secret')
    return items

my_list = ['a', 'b']
result = add_item_buggy(my_list)
print("Original modified:", my_list)

# With a tuple, the caller knows the data is safe
def process_coordinates(coords):
    # coords[0] = 0  # Would raise TypeError
    return coords

my_tuple = (10, 20)
result = process_coordinates(my_tuple)
print("Tuple unchanged:", my_tuple)
```

## Set Creation and the Uniqueness Guarantee

A set is an unordered collection of **unique** elements, created with curly braces `{}` or the `set()` constructor. Duplicates are automatically eliminated. Sets are also **unhashable** themselves (you cannot put a set inside another set), but all elements within a set must be hashable (immutable).

The uniqueness guarantee is extremely useful in data science for deduplication tasks. For example, finding the unique categories in a column, identifying unique customers, or removing duplicate entries from a list. Membership testing with `in` is O(1) on average for sets, compared to O(n) for lists — a huge performance win on large datasets.

```python
# Set creation and deduplication
numbers = [1, 2, 2, 3, 3, 3, 4, 5, 5]
unique_numbers = set(numbers)
print("Original list:", numbers)
print("Unique set:", unique_numbers)

# Direct set literal
categories = {"A", "B", "C", "A", "B"}
print("Categories:", categories)

# Fast membership testing
print("Is B a category?", "B" in categories)
print("Is Z a category?", "Z" in categories)
```

```python
# Deduplication in a data science context
user_ids = [101, 102, 103, 101, 104, 102, 105, 101]
unique_users = set(user_ids)
print("User IDs:", user_ids)
print("Unique users:", unique_users)
print("Number of unique users:", len(unique_users))

# Remove duplicates while preserving order (Python 3.7+)
seen = set()
ordered_unique = []
for uid in user_ids:
    if uid not in seen:
        ordered_unique.append(uid)
        seen.add(uid)
print("Ordered unique:", ordered_unique)
```

## Set Operations: Union, Intersection, Difference, Symmetric Difference

Sets support mathematical set operations that map directly to common data analysis tasks. The operators are: `|` for union (all elements from both sets), `&` for intersection (elements in both), `-` for difference (elements in the first but not the second), and `^` for symmetric difference (elements in either set but not both).

In data science, these operations are invaluable for comparing groups. For example: finding customers who purchased in both months (intersection), finding users who signed up but never purchased (difference), or combining two lists of tags (union). These operations are highly optimized and much faster than equivalent list comprehensions.

```python
# Set operations
power_users = {"Alice", "Bob", "Charlie", "Diana"}
active_users = {"Bob", "Diana", "Eve", "Frank"}

print("Power users:", power_users)
print("Active users:", active_users)

# Union: all users
all_users = power_users | active_users
print("Union (all):", all_users)

# Intersection: users in both groups
both = power_users & active_users
print("Intersection (both):", both)

# Difference: power users NOT active
power_only = power_users - active_users
print("Difference (power only):", power_only)

# Symmetric difference: in exactly one group
exactly_one = power_users ^ active_users
print("Symmetric diff:", exactly_one)
```

```python
# Set operations in data analysis
last_week = {"prod_a", "prod_b", "prod_c", "prod_d"}
this_week = {"prod_c", "prod_d", "prod_e", "prod_f"}

# Products launched both weeks
repeated = last_week & this_week
print("Repeated products:", repeated)

# New products this week (not sold last week)
new_products = this_week - last_week
print("New products:", new_products)

# Discontinued products (sold last week but not this week)
discontinued = last_week - this_week
print("Discontinued:", discontinued)
```

## Frozen Sets

A `frozenset` is an immutable version of a set. It supports all set operations but cannot be modified after creation — no `.add()`, `.remove()`, or `.discard()`. Because of its immutability, a `frozenset` is **hashable** and can be used as a dictionary key or as an element of another set.

Frozen sets are useful when you need to use a set as a dictionary key (e.g., grouping data by a set of tags) or when you want a constant set of values that should not be accidentally modified. They are also used in caching and memoization where the key needs to be hashable.

```python
# Frozenset creation
frozen = frozenset([1, 2, 3, 3, 4])
print("Frozenset:", frozen)
print("Type:", type(frozen))

# Frozensets are hashable — can be dictionary keys
group_a = frozenset({'feature_a', 'feature_b'})
group_b = frozenset({'feature_c', 'feature_d'})
feature_groups = {
    group_a: "Group A contains features for login",
    group_b: "Group B contains features for checkout"
}
for group, description in feature_groups.items():
    print(f"{group}: {description}")
```

```python
# Frozenset with set operations
base_tags = frozenset(['python', 'data-science', 'tutorial'])
extra_tags = frozenset(['python', 'machine-learning', 'beginner'])

# All operations work on frozensets
print("Union:", base_tags | extra_tags)
print("Intersection:", base_tags & extra_tags)
print("Difference:", base_tags - extra_tags)

# Frozenset inside a set (not possible with regular sets)
s = frozenset([1, 2])
t = frozenset([3, 4])
set_of_frozensets = {s, t}
print("Set of frozensets:", set_of_frozensets)
```
