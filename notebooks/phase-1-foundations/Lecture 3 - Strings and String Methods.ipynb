## String Immutability

Strings in Python are **immutable** — once a string is created, you cannot change individual characters in place. If you try `s[0] = 'X'`, Python raises a `TypeError`. Instead, string *methods* return a **new** string with the modification applied, leaving the original unchanged. This is a common source of bugs for beginners who forget to assign the result back to a variable.

Immutability has important benefits: strings can be used as keys in dictionaries, they are hashable, and they are safe to share across threads. In data science, you will often process text data (customer names, product descriptions, log files) by creating cleaned-up copies rather than modifying the originals — immutability encourages this safe pattern.

```python
# Strings are immutable — this will fail:
# s = "hello"
# s[0] = "H"  # TypeError: 'str' object does not support item assignment

# Correct: string methods return a new string
s = "hello"
s_capitalized = s.capitalize()
print("Original:", s)
print("New:", s_capitalized)

# Each method call creates a fresh string
name = "  ada lovelace  "
clean = name.strip().title()
print(f"Original: [{name}]")
print(f"Cleaned: [{clean}]")
```

```python
# Immutability error demonstration
# Trying to modify a string in place:
tweet = "i love data science"
# tweet[0] = "I"  # uncomment to see the error

# Correct approach: create a new string
tweet_fixed = tweet[0].upper() + tweet[1:]
print("Original:", tweet)
print("Fixed:", tweet_fixed)
```

## Common String Methods: .lower(), .upper(), .strip(), .split(), .join(), .replace()

Python strings come with a rich set of built-in methods. The most commonly used in data cleaning are: `.lower()` and `.upper()` for case normalization; `.strip()` (and its variants `.lstrip()` / `.rstrip()`) for removing whitespace from the edges; `.split()` for breaking a string into a list; `.join()` for assembling a list back into a string; and `.replace()` for substituting characters or substrings.

A powerful technique is **method chaining** — calling multiple methods on the same string in a single expression. Since each method returns a new string, you can chain them like `s.strip().lower().replace(' ', '_')`. This is idiomatic Python and widely used in data science pipelines to clean text data efficiently.

```python
# Case conversion and whitespace removal
raw = "  Hello WORLD!  "
print("lower():", raw.lower())
print("upper():", raw.upper())
print("strip():", raw.strip())
print("title():", raw.strip().title())

# Method chaining for data cleaning
dirty = "   DATA_SCIENCE   "
clean = dirty.strip().lower().replace('_', ' ')
print("Cleaned:", clean)
```

```python
# split() and join() — critical for CSV/text processing
csv_line = "Alice,25,Engineer,75000"
fields = csv_line.split(',')
print("split result:", fields)

# join() is called on the delimiter, takes a list of strings
tags = ['python', 'data-science', 'tutorial']
tag_string = ' | '.join(tags)
print("Joined tags:", tag_string)

# Practical: normalize user input
user_input = "  New York  "
normalized = user_input.strip().lower().replace(' ', '_')
print(f"Normalized city key: {normalized}")
```

```python
# replace() for cleaning text
review = "The product was good, but shipping was SLOW!!!"
clean_review = review.replace('!', '').replace('SLOW', 'slow')
print("Original:", review)
print("Cleaned:", clean_review)

# Remove multiple whitespace using split+join
messy = "Python    is   awesome"
clean = " ".join(messy.split())
print("Whitespace cleaned:", clean)
```

## Indexing (Positive and Negative)

Every character in a string has an index. Positive indices start at 0 for the first character and increase to the right. Negative indices start at -1 for the last character and decrease to the left. This dual indexing system lets you access characters from either end without needing to know the string length.

In data science, you will use indexing to extract specific characters from identifiers, pick the last letter of a code, or check the first character of a column value. For example, `user_id[-4:]` might give you the last 4 digits of a user ID — a quick way to extract a batch code or year.

```python
# Positive indexing (starts at 0)
word = "PYTHON"
print("Word:", word)
print("Index 0:", word[0])
print("Index 1:", word[1])
print("Index 5:", word[5])

# Negative indexing (starts at -1 for the last character)
print("Index -1:", word[-1])
print("Index -2:", word[-2])
print("Index -6:", word[-6])

# Practical: extract file extension
filename = "report_2024.csv"
extension = filename[-4:]
print(f"File: {filename}")
```

```python
# Accessing characters in a data science context
product_code = "DS-PY-101-B"
print("Product code:", product_code)
print("First char:", product_code[0])
print("Last char:", product_code[-1])
print("Third char:", product_code[2])

# Get the department code (first two characters)
dept = product_code[:2]
print("Department:", dept)

# IndexError when out of bounds
# print(product_code[20])  # uncomment to see IndexError
```

## Slicing Syntax [start:stop:step]

Slicing allows you to extract any contiguous (or non-contiguous) portion of a string. The syntax is `string[start:stop:step]`, where `start` is inclusive, `stop` is exclusive, and `step` is the increment. All three are optional. Omitting `start` defaults to the beginning, omitting `stop` defaults to the end, and omitting `step` defaults to 1.

Slicing is one of the most elegant features of Python. A negative step reverses direction, letting you reverse a string with `[::-1]`. In data science, slicing is essential for extracting date parts from ISO-formatted strings, trimming file extensions, or generating substrings for feature engineering.

```python
# Basic slicing examples
text = "Data Science"
print("Full string:", text)
print("text[0:4]:", text[0:4])
print("text[5:12]:", text[5:12])
print("text[:4]:", text[:4])
print("text[5:]:", text[5:])
print("text[:]:", text[:])  # full copy

# Using step
print("text[::2]:", text[::2])
print("text[::-1]:", text[::-1])  # reverse
```

```python
# Practical slicing for data cleaning
date_str = "2024-03-15"
year = date_str[:4]
month = date_str[5:7]
day = date_str[8:10]
print(f"Date: {year}-{month}-{day}")

# Extract domain from email
email = "alice@example.com"
at_pos = email.index('@')  # find the @ symbol
username = email[:at_pos]
domain = email[at_pos + 1:]
print(f"Username: {username}")
print(f"Domain: {domain}")

# Negative step to reverse
palindrome = "racecar"
print("Is palindrome?", palindrome == palindrome[::-1])
```

## f-string Formatting with Expressions

f-strings are not just for embedding simple variables — they can contain any valid Python expression, including arithmetic, function calls, and method calls. You can also use format specifiers after a colon to control how values are displayed: `:.2f` for two decimal places, `:.2%` for percentage, `:>10` for right-alignment within a width of 10, and many more.

In data science, f-strings are used everywhere: printing model metrics with controlled precision, formatting currency values, creating dynamic plot titles, and building report strings. They combine readability with power, making your output both informative and professional-looking.

```python
# Expressions inside f-strings
prices = [9.99, 15.49, 23.00]
print(f"Min: ${min(prices):.2f}, Max: ${max(prices):.2f}")

# Method calls inside f-strings
city = "  paris  "
print(f"City: {city.strip().title()}")

# Conditional expressions
score = 85
print(f"Result: {'Pass' if score >= 70 else 'Fail'}")
```

```python
# Format specifiers for alignment and padding
for name, price in [("Apple", 0.5), ("Banana", 0.25), ("Cherry", 0.75)]:
    print(f"{name:>8}  ${price:.2f}")

# Percentage formatting
accuracy = 0.8765
print(f"Model accuracy: {accuracy:.2%}")

# Large numbers with comma separator
population = 8_100_000_000
print(f"World population: {population:,}")
```
