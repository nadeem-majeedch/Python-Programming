## Variable Assignment and Naming Conventions

A variable is a name that refers to a value stored in memory. In Python, you create a variable with the assignment operator `=`. Python is dynamically typed, meaning a variable can hold any type of data, and the type can change over the life of the variable. This flexibility is great for rapid prototyping, but it also means you need to be careful about what type a variable holds at any given moment.

Python naming conventions (defined in PEP 8) recommend `snake_case` for variable names: lowercase words separated by underscores. Names should be descriptive but concise. Avoid single-letter names except for trivial loop counters (`i`, `j`). Also avoid reserved keywords like `for`, `if`, `while`, and built-in names like `list`, `dict`, `sum` — these will cause confusing errors.

```python
# Valid variable assignments
age = 28
temperature_celsius = 22.5
first_name = "Ada"
is_employed = True

# Print each variable to see its value
print("Age:", age)
print("Temperature:", temperature_celsius)
print("Name:", first_name)
print("Employed:", is_employed)

# Use type() to check the data type
print("type of age:", type(age))
print("type of temperature:", type(temperature_celsius))
```

```python
# Bad variable names to avoid
# l = [1, 2, 3]       # hard to read, shadows letter l
# O = 100             # looks like zero
# list = [1, 2, 3]    # shadows the built-in list() function

# Good variable names
customer_count = 143
total_revenue = 45999.95
is_subscribed = False

print(f"Customers: {customer_count}, Revenue: ${total_revenue}")
```

## Numeric Types: int vs float

Python has two primary numeric types: `int` (integer) and `float` (floating-point number). Integers are whole numbers with no decimal point, like `42` or `-7`. Floats are numbers with a decimal point, like `3.14` or `-0.001`. Unless you have a specific reason to use an integer, division in Python 3 always returns a float.

Be aware that floats have limited precision due to how computers represent them in binary. For example, `0.1 + 0.2` gives `0.30000000000000004`, not exactly `0.3`. This matters in data science when comparing floating-point numbers — never use `==` on floats; use a small tolerance instead. We will revisit this when we get to NumPy.

```python
# Integer operations
a = 10
b = 3
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)   # division always returns float
print("a // b =", a // b) # floor division
print("a % b =", a % b)   # modulo (remainder)
print("a ** b =", a ** b) # exponentiation
```

```python
# Float precision gotcha
print("0.1 + 0.2 =", 0.1 + 0.2)
print("Is 0.1 + 0.2 == 0.3?", 0.1 + 0.2 == 0.3)

# Better way to compare floats: use a tolerance
tolerance = 1e-9
result = 0.1 + 0.2
print("Close enough?", abs(result - 0.3) < tolerance)

# Type mixing: int + float => float
print("int + float:", 5 + 2.5)
```

## String Basics: Concatenation, Repetition, f-Strings

Strings can be joined using the `+` operator (concatenation) or repeated using the `*` operator. While `+` works, it becomes unwieldy when combining many values. A much cleaner approach is the **f-string** — a string literal prefixed with `f` or `F` that lets you embed expressions directly inside curly braces `{}`. Introduced in Python 3.6, f-strings are the modern, preferred way to build strings in Python.

In data science, you will use f-strings constantly — to format print statements, build file paths, create dynamic plot titles, and generate formatted report output. They are faster and more readable than older approaches like `%` formatting or `.format()`.

```python
# String concatenation with +
greeting = "Hello" + " " + "World"
print("Concatenation:", greeting)

# String repetition with *
separator = "=" * 30
print("Repetition:")
print(separator)
print("Chapter 1")
print(separator)
```

```python
# f-strings: embed variables and expressions
name = "Alice"
score = 95.5
print(f"Student: {name}, Score: {score}")

# f-strings can evaluate expressions inline
total = 100
count = 3
print(f"Average: {total / count:.2f}")  # :.2f means 2 decimal places

# f-strings in data science context
accuracy = 0.9342
model_name = 'Random Forest'
print(f"{model_name} accuracy: {accuracy:.2%}")  # :.2% formats as percentage
```

## Boolean Values and Comparison Operators

Booleans represent truth values: `True` and `False` (note the capital letters). They are the result of comparison operators like `==` (equal), `!=` (not equal), `<`, `>`, `<=`, `>=`. Booleans are the foundation of decision-making in code — they power `if` statements, `while` loops, and filter conditions in data analysis.

In data science, comparison operators are essential for filtering rows in a dataset. For example, `df[df['age'] > 30]` keeps only rows where the age column is greater than 30. Understanding how comparisons work at the fundamental level is critical before moving to pandas.

```python
# Comparison operators
print("5 == 5:", 5 == 5)
print("5 != 3:", 5 != 3)
print("10 > 7:", 10 > 7)
print("3 <= 3:", 3 <= 3)
print('"Python" == "python":', "Python" == "python")
print('"apple" < "banana":', "apple" < "banana")  # lexicographic comparison
```

```python
# Boolean logic in data filtering context
age = 25
salary = 60000
has_experience = True

# Combine comparisons with 'and' and 'or'
eligible = (age > 21) and (salary >= 50000)
print("Eligible for loan:", eligible)

hired = has_experience or (age < 30)
print("Candidate hired:", hired)

# Use not to invert a boolean
print("Not True:", not True)
print("Not False:", not False)
```

## Type Conversion (int(), float(), str(), bool())

Sometimes you need to explicitly convert a value from one type to another. Python provides built-in functions for this: `int()` converts to integer, `float()` to float, `str()` to string, and `bool()` to boolean. This is especially common in data science when reading data from files — numbers may arrive as strings and need converting before you can compute with them.

Not all conversions are valid. For example, `int('hello')` will raise a `ValueError`. Always validate or handle exceptions when converting user-supplied or file-supplied data. The `bool()` function has a special behavior: it returns `False` for `0`, `0.0`, `''` (empty string), `None`, and empty containers; everything else is `True`.

```python
# Converting between types
price_str = "49.99"
quantity_str = '3'

# Convert string to numeric types
price = float(price_str)
quantity = int(quantity_str)
total = price * quantity
print(f"Total: ${total:.2f}")

# Convert number back to string for display
total_str = str(total)
print("Type of total_str:", type(total_str))
```

```python
# bool() conversion rules
print("bool(0):", bool(0))
print("bool(1):", bool(1))
print("bool(\"\"):", bool(""))
print("bool(\"Python\"):", bool("Python"))
print("bool([]):", bool([]))
print("bool([1, 2]):", bool([1, 2]))
print("bool(None):", bool(None))
```
