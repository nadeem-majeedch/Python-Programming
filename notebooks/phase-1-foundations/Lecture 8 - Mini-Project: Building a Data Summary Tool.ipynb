## Putting It All Together: Lists of Dicts as a Mini DataFrame

Throughout Phase 1, you have learned about variables, data types, strings, lists, tuples, sets, dictionaries, conditionals, and loops. Now it is time to combine them all into a single, practical program. In this mini-project, you will build a **Data Summary Tool** that analyzes a small dataset stored as a list of dictionaries — exactly the same structure that pandas uses internally to represent a DataFrame.

Each dictionary in the list represents one record (like a row in a spreadsheet), with keys as column names and values as the data. This structure is intuitive, flexible, and widely used in real-world Python for representing structured data before loading it into specialized libraries. By the end of this lecture, you will have written a program that computes summary statistics, filters data, and generates a formatted report.

## Building the Dataset

We will create a dataset of employee records. Each record will be a dictionary with fields like `name`, `department`, `salary`, and `years_of_experience`. Storing data this way makes it easy to understand, modify, and extend. You can add more fields or more records without changing the code that processes them.

This dataset will serve as the input to our summary tool. In real data science work, this data would come from a CSV file or a database query, but the principles of analysis are exactly the same.

```python
# Step 1: Build the dataset (list of dicts)
employees = [
    {"name": "Alice", "dept": "Engineering", "salary": 95000, "years": 5},
    {"name": "Bob", "dept": "Marketing", "salary": 72000, "years": 3},
    {"name": "Charlie", "dept": "Engineering", "salary": 110000, "years": 8},
    {"name": "Diana", "dept": "Sales", "salary": 85000, "years": 4},
    {"name": "Eve", "dept": "Marketing", "salary": 68000, "years": 2},
    {"name": "Frank", "dept": "Engineering", "salary": 125000, "years": 12},
    {"name": "Grace", "dept": "Sales", "salary": 92000, "years": 6},
    {"name": "Hank", "dept": "Marketing", "salary": 77000, "years": 3},
]

print(f"Dataset loaded: {len(employees)} records")
print(f"Departments: {sorted(set(e['dept'] for e in employees))}")
print("\nFirst 3 records:")
for emp in employees[:3]:
    print(f"  {emp}")
```

## Computing Mean, Min, Max, and Count by Category

One of the most common data science tasks is computing summary statistics **grouped by category**. For example, you might want the average salary per department, or the minimum years of experience in each department. We will use a dictionary where each key is a department name and each value is a list of the relevant values, then compute statistics on each list.

This pattern — group, aggregate, report — is the foundation of tools like SQL's `GROUP BY` and pandas' `.groupby()`. Implementing it manually from scratch will give you a deep understanding of what those tools do under the hood.

```python
# Step 2: Compute summary statistics by department
from collections import defaultdict

# Group salaries by department
dept_salaries = defaultdict(list)
for emp in employees:
    dept_salaries[emp['dept']].append(emp['salary'])

print("Department Salary Summary")
print("=" * 50)

for dept, salaries in sorted(dept_salaries.items()):
    count = len(salaries)
    total = sum(salaries)
    mean = total / count
    min_sal = min(salaries)
    max_sal = max(salaries)
    print(f"{dept:15s}  Count: {count}  Mean: ${mean:,.0f}  "
          f"Min: ${min_sal:,}  Max: ${max_sal:,}")
```

```python
# Step 3: General-purpose group stats function
def group_stats(data, group_key, value_key):
    """Group data by group_key and compute stats on value_key."""
    grouped = defaultdict(list)
    for record in data:
        grouped[record[group_key]].append(record[value_key])

    results = {}
    for group, values in grouped.items():
        results[group] = {
            'count': len(values),
            'mean': sum(values) / len(values),
            'min': min(values),
            'max': max(values),
            'total': sum(values),
        }
    return results


# Use the function
stats = group_stats(employees, 'dept', 'salary')
print("Engineering stats:", stats["Engineering"])
print("Marketing stats:", stats["Marketing"])
```

## Filtering Records with Conditionals

Another fundamental data operation is **filtering** — selecting only the records that meet certain criteria. You have already learned about `if` statements and comparison operators. Now you will apply them to filter our employee dataset. For example, you might want only employees with a salary above a threshold, or only those in a specific department.

Filtering is done with a list comprehension or a for loop with an `if` condition. This is exactly what pandas does internally when you write `df[df['salary'] > 90000]`. Building the filter manually reinforces how the operation works at a fundamental level.

```python
# Step 4: Filter records with conditionals
# Filter: employees with salary > 90000
high_earners = [emp for emp in employees if emp["salary"] > 90000]
print(f"High earners (salary > $90k): {len(high_earners)} employees")
for emp in high_earners:
    print(f"  {emp['name']:10s}  {emp['dept']:15s}  ${emp['salary']:,}")

# Filter: employees with 5+ years experience
senior = [emp for emp in employees if emp["years"] >= 5]
print(f"\nSenior employees (5+ years): {len(senior)}")
for emp in senior:
    print(f"  {emp['name']:10s}  {emp['dept']:15s}  {emp['years']} years")
```

```python
# Step 5: Multi-condition filtering
# Filter: Engineering dept with salary > 100k
eng_high = [
    emp for emp in employees
    if emp['dept'] == 'Engineering' and emp['salary'] > 100000
]
print("Engineering high earners:")
for emp in eng_high:
    print(f"  {emp['name']} - ${emp['salary']:,}")

# Filter: Marketing or Sales with < 5 years experience
junior_business = [
    emp for emp in employees
    if emp['dept'] in ('Marketing', 'Sales') and emp['years'] < 5
]
print("\nJunior Marketing/Sales employees:")
for emp in junior_business:
    print(f"  {emp['name']} - {emp['dept']} - {emp['years']} years")
```

## Formatting a Clean Text Report

The final step is to present our findings in a well-formatted report. A good data science report is readable, structured, and contains all the key information without unnecessary detail. We will use f-strings with alignment specifiers, separators, and clear section headings to build a professional-looking output.

This skill — presenting results clearly — is just as important as the analysis itself. In practice, you would write this report to a file or include it in an email. The formatting techniques you use here (alignment, separators, grouping) apply equally to generating PDF reports, HTML tables, or even dashboard labels.

```python
# Step 6: Generate a complete formatted report
def generate_report(data):
    lines = []
    lines.append("=" * 60)
    lines.append("EMPLOYEE DATA SUMMARY REPORT")
    lines.append("=" * 60)

    # Section 1: Overview
    lines.append(f"\nTotal employees: {len(data)}")
    departments = sorted(set(e["dept"] for e in data))
    lines.append(f"Departments: {", ".join(departments)}")
    total_salary = sum(e['salary'] for e in data)
    lines.append(f'Total salary budget: ${total_salary:,}')
    avg_salary = total_salary / len(data)
    lines.append(f'Overall average salary: ${avg_salary:,.0f}')

    # Section 2: By-department breakdown
    lines.append("-" * 60)
    lines.append("BREAKDOWN BY DEPARTMENT")
    lines.append("-" * 60)

    stats = group_stats(data, "dept", "salary")
    for dept in sorted(stats.keys()):
        s = stats[dept]
        lines.append(
            f"{dept:15s} | Count: {s['count']}  "
            f"Avg: ${s['mean']:>8,.0f}  "
            f"Min: ${s['min']:>6,}  Max: ${s['max']:>6,}"
        )

    # Section 3: High earners
    lines.append("-" * 60)
    lines.append("HIGH EARNERS (Salary > $90,000)")
    lines.append("-" * 60)
    high = [e for e in data if e["salary"] > 90000]
    for e in sorted(high, key=lambda x: x['salary'], reverse=True):
        lines.append(
            f"  {e['name']:12s}  {e['dept']:15s}  ${e['salary']:,}"
        )

    lines.append("=" * 60)
    lines.append("END OF REPORT")
    lines.append("=" * 60)
    return '\n'.join(lines)


# Generate and print the report
report = generate_report(employees)
print(report)
```

```python
# Step 7: Save report to a file
report = generate_report(employees)
with open('employee_summary_report.txt', 'w') as f:
    f.write(report)
print("Report saved to employee_summary_report.txt")

# Bonus: Filter by years of experience and compute stats
def filter_and_report(data, min_years):
    filtered = [e for e in data if e['years'] >= min_years]
    print(f'\nEmployees with {min_years}+ years: {len(filtered)}')
    if filtered:
        avg_sal = sum(e['salary'] for e in filtered) / len(filtered)
        print(f'Average salary: ${avg_sal:,.0f}')
        depts = {e['dept'] for e in filtered}
        print(f'Departments represented: {depts}')


filter_and_report(employees, 5)
filter_and_report(employees, 10)
```
