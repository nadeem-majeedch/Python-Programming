## List Creation and Heterogeneous Elements

A list is an ordered, mutable collection of elements enclosed in square brackets `[]`. Unlike arrays in many other languages, Python lists can hold elements of **different types** — you can mix integers, strings, floats, and even other lists in a single list. This flexibility makes lists ideal for storing heterogeneous data like a row from a spreadsheet.

Lists are the closest native Python data structure to a data frame row or a CSV record. In data science, you will often use lists to hold column names, store raw data before converting it to a NumPy array or pandas DataFrame, or accumulate results during iteration.

```python
# Creating lists with mixed types
empty_list = []
numbers = [1, 2, 3, 4, 5]
mixed = [42, "hello", 3.14, True]
print("Empty:", empty_list)
print("Numbers:", numbers)
print("Mixed:", mixed)

# List from a range (common in data processing)
indices = list(range(10))
print("Indices:", indices)
```

```python
# Lists as data records
# Think of each list as one row of a dataset
record_1 = ["Alice", 28, "Engineer", 75000]
record_2 = ["Bob", 35, "Data Scientist", 95000]
record_3 = ["Charlie", 22, "Intern", 45000]

all_records = [record_1, record_2, record_3]
print("All records:", all_records)

# Check the length
print("Number of records:", len(all_records))
print("Fields in first record:", len(record_1))
```

## Indexing, Slicing, and Nested Lists

Lists support the same indexing and slicing syntax as strings: `my_list[0]` gives the first element, `my_list[-1]` gives the last, and `my_list[1:4]` gives elements at indices 1, 2, and 3. But unlike strings, lists are mutable — you can assign to an index: `my_list[0] = new_value`. This mutability is what makes lists suitable for building and modifying collections of data.

Nested lists (lists within lists) are useful for representing 2D data like spreadsheets or matrices. Accessing elements in nested lists uses multiple indices: `matrix[row][col]`. In data science, you might use nested lists to hold a small dataset before converting it to a NumPy array for numerical computation.

```python
# Indexing and slicing
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print("First:", fruits[0])
print("Last:", fruits[-1])
print("First three:", fruits[:3])
print("Every other:", fruits[::2])

# Lists are mutable — assign to an index
fruits[1] = "blueberry"
print("After change:", fruits)
```

```python
# Nested lists (2D data)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Matrix:", matrix)
print("Row 0:", matrix[0])
print("Element [1][2]:", matrix[1][2])

# Nested list as a mini spreadsheet
data = [
    ["Name", "Age", "Salary"],
    ["Alice", 28, 75000],
    ["Bob", 35, 95000]
]
print("Column names:", data[0])
print("First data row:", data[1])
print("Bob's salary:", data[2][2])
```

## List Methods: .append(), .extend(), .insert(), .remove(), .pop(), .sort()

Lists have a rich set of methods for modification. The most important for data science are: `.append()` to add a single element to the end; `.extend()` to add all elements from another iterable; `.insert()` to add at a specific position; `.remove()` to delete the first occurrence of a value; `.pop()` to remove and return an element by index (default is the last element); and `.sort()` to sort in place.

Note that methods like `.append()` and `.sort()` modify the list **in place** and return `None`. This is a common pitfall: `my_list = my_list.append(x)` will set `my_list` to `None`. Always call these methods without assignment.

```python
# append() and extend()
stack = []
stack.append(10)
stack.append(20)
stack.append(30)
print("After appends:", stack)

more = [40, 50]
stack.extend(more)
print("After extend:", stack)

# insert() and remove()
stack.insert(0, 5)
print("After insert at 0:", stack)
stack.remove(30)
print("After remove(30):", stack)
```

```python
# pop() and sort()
tasks = [3, 1, 4, 1, 5, 9, 2, 6]
print("Original:", tasks)

# Sort in place (modifies the list)
tasks.sort()
print("Sorted:", tasks)

# Sort in reverse
tasks.sort(reverse=True)
print("Reverse sorted:", tasks)

# Pop from the end (like a stack)
last = tasks.pop()
print("Popped:", last)
print("Remaining:", tasks)

# sorted() returns a NEW list (does not modify original)
original = [3, 1, 4]
sorted_copy = sorted(original)
print("Original unchanged:", original)
print("Sorted copy:", sorted_copy)
```

## The `in` Operator and List Membership

The `in` operator checks whether a value exists in a list, returning `True` or `False`. It is a clean and readable way to test membership: `if 'apple' in fruits:`. The `not in` variant checks for absence. Under the hood, Python scans the list from beginning to end (O(n) time complexity), so for very large lists this is slower than using a set.

In data science, `in` is often used in filtering — for example, keeping only rows where a category value is in a whitelist: `if category in ['A', 'B', 'C']:`. It is also useful for data validation, such as checking whether a required column name exists before processing.

```python
# The in operator
colors = ["red", "green", "blue"]
print("red in colors:", "red" in colors)
print("yellow in colors:", "yellow" in colors)
print("yellow not in colors:", "yellow" not in colors)

# Filtering with in
allowed_categories = ["A", "B", "C"]
record = "B"
if record in allowed_categories:
    print(f"Category {record} is allowed")

# in with nested structures
dataset = [
    {"id": 1, "status": "active"},
    {"id": 2, "status": "inactive"}
]
# Check if any record has a certain value
statuses = [r["status"] for r in dataset]
print("active present:", "active" in statuses)
```

## Shallow Copies vs Deep Copies with copy()

When you assign a list to a new variable with `new_list = old_list`, you are NOT creating a copy — both variables point to the same object in memory. Modifying one will affect the other. To create an actual copy, you can use `list.copy()` or `old_list[:]`, which creates a **shallow copy** — a new list object whose elements are references to the same objects as the original.

For nested lists, a shallow copy is often insufficient because modifying a nested element (like a sublist) will still affect both the original and the copy. To fully decouple nested structures, you need a **deep copy** via `copy.deepcopy()`. In data science, understanding copies is critical when preprocessing data — you want to avoid accidentally mutating your original data.

```python
# Assignment vs copy
original = [1, 2, 3]
assigned = original  # both point to the same list
assigned.append(4)
print("Original after assign-append:", original)

# Shallow copy with .copy()
original = [1, 2, 3]
shallow = original.copy()
shallow.append(4)
print("Original after copy-append:", original)
print("Shallow copy:", shallow)
```

```python
# Shallow copy pitfall with nested lists
original = [[1, 2], [3, 4]]
shallow = original.copy()
shallow[0].append(99)
print("Original:", original)
print("Shallow:", shallow)

# Deep copy to the rescue
import copy
original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)
deep[0].append(99)
print("Original after deep copy mod:", original)
print("Deep copy:", deep)
```
